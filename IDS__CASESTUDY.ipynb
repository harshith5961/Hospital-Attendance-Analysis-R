### Data Cleaning and Waiting Period Calculation

````r
names(appointments)[
  names(appointments) %in% c("No-show", "No.show")
] <- "verification"

stopifnot("verification" %in% names(appointments))

appointments_clean <- appointments[appointments$Age >= 0, ]

appointments_clean$ScheduledDay <- as.Date(
  substr(as.character(appointments_clean$ScheduledDay), 1, 10)
)

appointments_clean$AppointmentDay <- as.Date(
  substr(as.character(appointments_clean$AppointmentDay), 1, 10)
)

appointments_clean$WaitingDays <- as.integer(
  appointments_clean$AppointmentDay -
    appointments_clean$ScheduledDay
)

cat("Appointments scheduled before booking date:",
    sum(appointments_clean$WaitingDays < 0, na.rm = TRUE), "\n")

appointments_clean <- appointments_clean[
  !is.na(appointments_clean$WaitingDays) &
    appointments_clean$WaitingDays >= 0,
]

appointments_clean$Gender <- factor(appointments_clean$Gender)

attendance <- as.character(appointments_clean$verification)
attendance[attendance == "No"] <- "Attended"
attendance[attendance == "Yes"] <- "Missed"

stopifnot(all(attendance %in% c("Attended", "Missed")))

appointments_clean$verification <- factor(
  attendance,
  levels = c("Attended", "Missed")
)

dim(appointments_clean)
summary(appointments_clean$WaitingDays)
table(appointments_clean$verification)
````

Training and Testing Data Split

````r
set.seed(33)

patient_ids <- unique(appointments_clean$PatientId)

train_patient_ids <- sample(
  patient_ids,
  size = floor(0.8 * length(patient_ids))
)

is_train <- appointments_clean$PatientId %in% train_patient_ids

train_data <- appointments_clean[is_train, ]
test_data <- appointments_clean[!is_train, ]

cat("Training appointments:", nrow(train_data), "\n")
cat("Testing appointments:", nrow(test_data), "\n")

table(train_data$verification)
table(test_data$verification)

cat("Patients shared between sets:",
    length(intersect(
      train_data$PatientId,
      test_data$PatientId
    )), "\n")
````

````r
features <- c(
  "Age", "Gender", "Scholarship", "Hipertension",
  "Diabetes", "Alcoholism", "Handcap",
  "SMS_received", "WaitingDays"
)

set.seed(33)

training_patient_ids <- unique(train_data$PatientId)

fit_patient_ids <- sample(
  training_patient_ids,
  size = floor(0.8 * length(training_patient_ids))
)

is_fit <- train_data$PatientId %in% fit_patient_ids

fit_data <- train_data[is_fit, ]
validation_data <- train_data[!is_fit, ]

fit_model_data <- fit_data[, c(features, "verification")]
validation_model_data <- validation_data[, c(features, "verification")]
test_model_data <- test_data[, c(features, "verification")]

cat("Training:", nrow(fit_model_data), "\n")
cat("Validation:", nrow(validation_model_data), "\n")
cat("Testing:", nrow(test_model_data), "\n")
````

Random Forest Model Training

````r
library(ranger)

rf_model <- ranger(
  verification ~ .,
  data = fit_model_data,
  probability = TRUE,
  num.trees = 300,
  mtry = 3,
  min.node.size = 20,
  importance = "impurity",
  num.threads = 2,
  seed = 33
)

print(rf_model)
````

Random Forest Test Evaluation

````r
rf_probabilities <- predict(
  rf_model,
  data = test_model_data
)$predictions

rf_missed_probability <- rf_probabilities[, "Missed"]

rf_prediction <- factor(
  ifelse(rf_missed_probability >= 0.5, "Missed", "Attended"),
  levels = c("Attended", "Missed")
)

rf_confusion <- table(
  Actual = test_model_data$verification,
  Predicted = rf_prediction
)

print(rf_confusion)

tp <- rf_confusion["Missed", "Missed"]
tn <- rf_confusion["Attended", "Attended"]
fp <- rf_confusion["Attended", "Missed"]
fn <- rf_confusion["Missed", "Attended"]

accuracy <- (tp + tn) / sum(rf_confusion)
precision <- ifelse(tp + fp > 0, tp / (tp + fp), 0)
recall <- tp / (tp + fn)
f1 <- ifelse(precision + recall > 0,
             2 * precision * recall / (precision + recall), 0)

actual_missed <- as.integer(test_model_data$verification == "Missed")
baseline_probability <- mean(train_model_data$verification == "Missed")

results <- data.frame(
  Metric = c(
    "Accuracy", "Missed precision", "Missed recall", "Missed F1",
    "Always-attend accuracy", "Model Brier score", "Baseline Brier score"
  ),
  Value = c(
    accuracy, precision, recall, f1,
    mean(actual_missed == 0),
    mean((rf_missed_probability - actual_missed)^2),
    mean((baseline_probability - actual_missed)^2)
  )
)

results$Value <- round(results$Value, 4)
results
````

XGBoost Model Training

````r
x_fit <- model.matrix(
  ~ . - 1,
  data = fit_model_data[, features]
)

x_validation <- model.matrix(
  ~ . - 1,
  data = validation_model_data[, features]
)

stopifnot(identical(colnames(x_fit), colnames(x_validation)))

y_fit <- as.integer(fit_model_data$verification == "Missed")
y_validation <- as.integer(
  validation_model_data$verification == "Missed"
)

d_fit <- xgboost::xgb.DMatrix(x_fit, label = y_fit)
d_validation <- xgboost::xgb.DMatrix(
  x_validation,
  label = y_validation
)

set.seed(33)

xgb_model <- xgboost::xgb.train(
  params = list(
    objective = "binary:logistic",
    eval_metric = "logloss",
    eta = 0.05,
    max_depth = 4,
    min_child_weight = 10,
    subsample = 0.8,
    colsample_bytree = 0.8,
    nthread = 2,
    seed = 33
  ),
  data = d_fit,
  nrounds = 500,
  watchlist = list(
    train = d_fit,
    validation = d_validation
  ),
  early_stopping_rounds = 30,
  verbose = 1
)
````

Validation Predictions

validation threshold using F1-score

````r
calculate_metrics <- function(actual, probability, threshold) {
  predicted <- as.integer(probability >= threshold)

  tp <- sum(actual == 1 & predicted == 1)
  tn <- sum(actual == 0 & predicted == 0)
  fp <- sum(actual == 0 & predicted == 1)
  fn <- sum(actual == 1 & predicted == 0)

  precision <- if (tp + fp > 0) tp / (tp + fp) else 0
  recall <- if (tp + fn > 0) tp / (tp + fn) else 0

  f1 <- if (precision + recall > 0) {
    2 * precision * recall / (precision + recall)
  } else 0

  data.frame(
    Threshold = threshold,
    Accuracy = (tp + tn) / length(actual),
    Precision = precision,
    Recall = recall,
    F1 = f1
  )
}

find_threshold <- function(actual, probability) {
  scores <- do.call(
    rbind,
    lapply(seq(0.01, 0.99, by = 0.01), function(t) {
      calculate_metrics(actual, probability, t)
    })
  )

  scores[which.max(scores$F1), , drop = FALSE]
}

validation_actual <- as.integer(
  validation_model_data$verification == "Missed"
)

rf_best <- find_threshold(
  validation_actual,
  rf_validation_prob
)

xgb_best <- find_threshold(
  validation_actual,
  xgb_validation_prob
)

rf_threshold <- rf_best$Threshold
xgb_threshold <- xgb_best$Threshold

validation_results <- rbind(
  cbind(Model = "Random Forest", rf_best),
  cbind(Model = "XGBoost", xgb_best)
)

print(validation_results)
````

Test Evaluation and Model Comparison

````r
rf_test_prob <- predict(
  rf_model,
  data = test_model_data
)$predictions[, "Missed"]

x_test <- model.matrix(
  ~ . - 1,
  data = test_model_data[, features]
)

stopifnot(identical(colnames(x_fit), colnames(x_test)))

d_test <- xgboost::xgb.DMatrix(x_test)

xgb_test_prob <- predict(xgb_model, d_test)

test_actual <- as.integer(
  test_model_data$verification == "Missed"
)

rf_test_prediction <- factor(
  ifelse(rf_test_prob >= rf_threshold, "Missed", "Attended"),
  levels = c("Attended", "Missed")
)

xgb_test_prediction <- factor(
  ifelse(xgb_test_prob >= xgb_threshold, "Missed", "Attended"),
  levels = c("Attended", "Missed")
)

cat("Random Forest Confusion Matrix\n")
print(table(
  Actual = test_model_data$verification,
  Predicted = rf_test_prediction
))

cat("\nXGBoost Confusion Matrix\n")
print(table(
  Actual = test_model_data$verification,
  Predicted = xgb_test_prediction
))

comparison <- rbind(
  cbind(
    Model = "Random Forest",
    calculate_metrics(test_actual, rf_test_prob, rf_threshold)
  ),
  cbind(
    Model = "XGBoost",
    calculate_metrics(test_actual, xgb_test_prob, xgb_threshold)
  )
)

comparison$BrierScore <- c(
  mean((rf_test_prob - test_actual)^2),
  mean((xgb_test_prob - test_actual)^2)
)

numeric_columns <- vapply(comparison, is.numeric, logical(1))

comparison[numeric_columns] <- lapply(
  comparison[numeric_columns],
  round,
  digits = 4
)

print(comparison)
````

confusion matrix plots

````r
plot_confusion <- function(actual, predicted, title) {
  cm <- table(
    Actual = factor(actual, levels = c("Attended", "Missed")),
    Predicted = factor(predicted, levels = c("Attended", "Missed"))
  )

  plot(
    NA,
    xlim = c(0.5, 2.5),
    ylim = c(0.5, 2.5),
    xlab = "Predicted",
    ylab = "Actual",
    main = title,
    axes = FALSE,
    asp = 1
  )

  for (i in 1:2) {
    for (j in 1:2) {
      y <- 3 - i
      rect(
        j - 0.5, y - 0.5,
        j + 0.5, y + 0.5,
        col = if (i == j) "lightblue" else "mistyrose",
        border = "white"
      )
      text(j, y, labels = cm[i, j], cex = 1.3)
    }
  }

  axis(1, at = 1:2, labels = colnames(cm))
  axis(2, at = c(2, 1), labels = rownames(cm), las = 1)
}

old_par <- par(mfrow = c(1, 2))

plot_confusion(
  test_model_data$verification,
  rf_test_prediction,
  "Random Forest"
)

plot_confusion(
  test_model_data$verification,
  xgb_test_prediction,
  "XGBoost"
)

par(old_par)
````

````r
report_dir <- "/content/hospital_attendance_report"
dir.create(report_dir, showWarnings = FALSE)

binary_features <- c(
  "Scholarship", "Hipertension", "Diabetes",
  "Alcoholism", "SMS_received"
)

if (!all(vapply(
  new_patient[, binary_features, drop = FALSE],
  function(x) all(x %in% c(0, 1)),
  logical(1)
))) {
  stop("New patient binary fields must be 0 or 1. Correct the patient details and rerun the prediction cell first.")
}

report_data <- list(
  original_rows = nrow(appointments),
  original_columns = ncol(appointments),
  cleaned_rows = nrow(appointments_clean),
  training_rows = nrow(fit_model_data),
  validation_rows = nrow(validation_model_data),
  testing_rows = nrow(test_model_data),
  clean_data = appointments_clean,
  comparison = comparison,
  validation_results = validation_results,
  actual = test_model_data$verification,
  rf_prediction = rf_test_prediction,
  xgb_prediction = xgb_test_prediction,
  rf_probability = rf_test_prob,
  xgb_probability = xgb_test_prob,
  baseline_probability = mean(
    fit_model_data$verification == "Missed"
  ),
  rf_importance = rf_model$variable.importance,
  xgb_importance = xgboost::xgb.importance(
    feature_names = colnames(x_fit),
    model = xgb_model
  ),
  xgb_log = as.data.frame(xgb_model$evaluation_log),
  new_patient = new_patient,
  new_patient_results = new_patient_results,
  rf_threshold = rf_threshold,
  xgb_threshold = xgb_threshold,
  features = features,
  patient_overlap = c(
    Training_Validation = length(intersect(
      fit_data$PatientId, validation_data$PatientId
    )),
    Training_Testing = length(intersect(
      fit_data$PatientId, test_data$PatientId
    )),
    Validation_Testing = length(intersect(
      validation_data$PatientId, test_data$PatientId
    ))
  ),
  software_environment = capture.output(sessionInfo())
)

stopifnot(
  all(report_data$patient_overlap == 0),
  length(report_data$actual) == length(report_data$rf_probability),
  length(report_data$actual) == length(report_data$xgb_probability)
)

saveRDS(
  report_data,
  file.path(report_dir, "report_data.rds")
)

report_text <- r"---(
---
title: "Hospital Appointment Management and Patient Attendance Analysis Using R"
author: "Harshit Srinivasa Prasad"
date: "`r format(Sys.Date(), '%d %B %Y')`"
output:
  html_document:
    toc: true
    toc_depth: 2
    number_sections: true
    theme: flatly
    highlight: tango
    self_contained: true
---

```{r setup, include=FALSE}
knitr::opts_chunk$set(
  echo = FALSE,
  warning = FALSE,
  message = FALSE,
  fig.width = 9,
  fig.height = 5,
  fig.align = "center"
)

report <- readRDS("report_data.rds")
dat <- report$clean_data
missed <- as.integer(dat$verification == "Missed")
```

# Abstract

This project analyses hospital appointment records and estimates the
probabilities of attending and missing an appointment. Data preparation
includes checking missing values and duplicates, removing invalid ages,
converting dates, and calculating waiting days. Random Forest and XGBoost
learn from patient and appointment characteristics. Patients remain
separate across training, validation, and testing partitions.
Validation F1-score determines follow-up thresholds, while test outcomes
support model evaluation. A fictional patient example demonstrates the
final probability output. The results support a prototype for prioritising
reminders, rather than a guarantee of individual attendance.

# Introduction

Missed appointments can leave scheduled capacity unused and complicate
hospital planning. Historical records can reveal attendance patterns
and provide information for estimating the risk of a missed appointment.

This project implements attendance analysis and prediction in R.
Appointment booking, cancellation, rescheduling, and automatic SMS delivery
are outside its implemented scope.

# Problem Statement

How can historical patient and appointment information help identify
appointments at higher risk of being missed?

The target has two outcomes: **Attended** and **Missed**.
Both models provide an estimated probability of missing, from which
the attending probability is calculated.

# Objectives

- Prepare historical appointment records for analysis.
- Analyse attendance patterns across patient groups.
- Train Random Forest and XGBoost models.
- Select follow-up thresholds using validation records.
- Compare test performance with a simple baseline.
- Display attending and missing probabilities for a new patient.

# Dataset Description

The working file is `KaggleV2-May-2016.csv`, from the
Medical Appointment No Shows dataset. Each row represents an appointment.
A patient may have multiple appointments.

```{r dataset-summary}
knitr::kable(data.frame(
  Description = c(
    "Imported appointments",
    "Original columns",
    "Cleaned appointments",
    "Rows removed during cleaning"
  ),
  Value = c(
    report$original_rows,
    report$original_columns,
    report$cleaned_rows,
    report$original_rows - report$cleaned_rows
  )
))
```

These counts describe the file actually imported for this experiment.
A partial working file should not be described as the complete source
dataset.

## Column Descriptions

```{r dictionary}
dictionary <- data.frame(
  Column = c(
    "PatientId", "AppointmentID", "Gender",
    "ScheduledDay", "AppointmentDay", "Age",
    "Neighbourhood", "Scholarship", "Hipertension",
    "Diabetes", "Alcoholism", "Handcap",
    "SMS_received", "verification", "WaitingDays"
  ),
  Description = c(
    "Patient identifier used to separate patients across partitions",
    "Appointment identifier",
    "Recorded gender",
    "Date the appointment was booked",
    "Date of the appointment",
    "Age in years",
    "Recorded neighbourhood",
    "Welfare programme indicator",
    "Recorded hypertension indicator",
    "Recorded diabetes indicator",
    "Recorded alcoholism indicator",
    "Recorded disability category",
    "Recorded SMS reminder indicator",
    "Attendance target renamed from No-show",
    "Calendar days between booking and appointment"
  )
)

knitr::kable(dictionary)
```

The original target uses **Yes for missed appointments** and
**No for attended appointments**. Cleaning replaces these with
the explicit labels **Missed** and **Attended**.

The SMS indicator does not establish that the patient read the message.

# Data Preparation

The preparation workflow includes:

1. Inspecting data types, missing values, duplicate rows, and ages.
2. Renaming the target column to `verification`.
3. Removing records with negative ages.
4. Converting booking and appointment values to calendar dates.
5. Calculating waiting days.
6. Excluding negative or missing waiting periods.
7. Converting gender and the target into factors.

Age zero remains valid for infants. Multiple appointments for the same
patient are retained. Waiting days are based on calendar dates because
the appointment field does not provide a reliable appointment time.

## Missing Values After Cleaning

```{r missing-values}
knitr::kable(data.frame(
  Column = names(dat),
  Missing_Values = as.integer(colSums(is.na(dat)))
))
```

## Waiting Period Summary

```{r waiting-summary}
knitr::kable(data.frame(
  Statistic = c("Minimum", "Median", "Mean", "Maximum"),
  Waiting_Days = c(
    min(dat$WaitingDays),
    median(dat$WaitingDays),
    mean(dat$WaitingDays),
    max(dat$WaitingDays)
  )
), digits = 2)
```

# Exploratory Analysis

The following summaries describe the entire cleaned working dataset.
They show associations and are not used to select model thresholds.

## Overall Attendance

```{r overall-attendance}
counts <- table(dat$verification)

attendance_table <- data.frame(
  Outcome = names(counts),
  Appointments = as.integer(counts),
  Percentage = 100 * as.integer(counts) / sum(counts)
)

knitr::kable(attendance_table, digits = 2)

barplot(
  counts,
  col = c("seagreen", "tomato"),
  main = "Overall Appointment Attendance",
  ylab = "Number of Appointments"
)
```

The cleaned dataset has a missed-appointment rate of
**`r round(100 * mean(missed), 2)`%**.

Unequal outcome frequencies make accuracy alone insufficient for judging
whether a model identifies missed appointments effectively.

## Attendance Patterns by Group

```{r grouped-analysis, fig.height=8}
age_group <- cut(
  dat$Age,
  breaks = c(-Inf, 17, 39, 59, Inf),
  labels = c("0-17", "18-39", "40-59", "60+")
)

waiting_group <- cut(
  dat$WaitingDays,
  breaks = c(-1, 0, 7, 30, Inf),
  labels = c("Same day", "1-7 days", "8-30 days", "31+ days")
)

rate_plot <- function(group, title) {
  rates <- 100 * tapply(missed, group, mean)

  barplot(
    rates,
    col = "steelblue",
    ylim = c(0, max(rates, na.rm = TRUE) + 10),
    main = title,
    ylab = "Missed Appointments (%)"
  )
}

old_par <- par(mfrow = c(2, 2))

rate_plot(age_group, "Missed Rate by Age")
rate_plot(dat$Gender, "Missed Rate by Gender")
rate_plot(waiting_group, "Missed Rate by Waiting Period")

rate_plot(
  factor(
    dat$SMS_received,
    levels = c(0, 1),
    labels = c("No SMS", "SMS")
  ),
  "Missed Rate by SMS Indicator"
)

par(old_par)
```

Percentages allow comparison across groups of different sizes.
Differences between SMS groups may reflect other characteristics,
including waiting periods. They do not prove an effect of reminders.

# Feature Selection

```{r features}
knitr::kable(data.frame(
  Predictor = report$features
))
```

Patient identifiers, appointment identifiers, neighbourhood, and raw dates
are excluded from the selected predictor set. Waiting days captures
the interval between the two dates.

Random Forest uses the prepared data frame. XGBoost uses a numeric
matrix with gender indicator columns. The target encoding for XGBoost is
**0 = Attended** and **1 = Missed**.

# Training Validation and Testing

The workflow first allocates approximately 80% of patients to training
and 20% to testing. It then reserves approximately 20% of training
patients for validation.

Appointment proportions can differ from patient proportions because
patients have different numbers of appointments.

```{r partitions}
knitr::kable(data.frame(
  Partition = c("Model training", "Validation", "Testing"),
  Appointments = c(
    report$training_rows,
    report$validation_rows,
    report$testing_rows
  )
))

knitr::kable(data.frame(
  Partition_Pair = names(report$patient_overlap),
  Shared_Patients = as.integer(report$patient_overlap)
))
```

Training fits the models. Validation supports early stopping and threshold
selection. Test outcomes evaluate the fixed models and thresholds.

# Model Development

## Random Forest

The Random Forest implementation uses `ranger` with probability estimation
enabled. It combines predictions from multiple trees.

```{r rf-settings}
knitr::kable(data.frame(
  Parameter = c(
    "Number of trees",
    "Candidate predictors per split",
    "Minimum node size",
    "Random seed"
  ),
  Value = c("300", "3", "20", "33")
))
```

The model uses only the model-training partition.

## XGBoost

XGBoost builds boosted trees sequentially. The binary logistic objective
produces estimated missed-appointment probabilities.

```{r xgb-settings}
knitr::kable(data.frame(
  Parameter = c(
    "Objective", "Evaluation metric", "Learning rate",
    "Maximum depth", "Minimum child weight",
    "Row sampling", "Column sampling",
    "Maximum rounds", "Early stopping patience", "Seed"
  ),
  Value = c(
    "binary:logistic", "Log loss", "0.05",
    "4", "10", "0.8", "0.8", "500", "30", "33"
  )
))
```

## XGBoost Learning Curve

```{r learning-curve}
logs <- report$xgb_log

train_columns <- grep(
  "train.*logloss",
  names(logs),
  value = TRUE
)

validation_columns <- grep(
  "validation.*logloss",
  names(logs),
  value = TRUE
)

if (length(train_columns) > 0 && length(validation_columns) > 0) {
  matplot(
    seq_len(nrow(logs)),
    cbind(
      logs[[train_columns[1]]],
      logs[[validation_columns[1]]]
    ),
    type = "l",
    lty = 1,
    col = c("steelblue", "tomato"),
    xlab = "Boosting Round",
    ylab = "Log Loss",
    main = "XGBoost Training and Validation Loss"
  )

  legend(
    "topright",
    legend = c("Training", "Validation"),
    col = c("steelblue", "tomato"),
    lty = 1,
    bty = "n"
  )
} else {
  cat("Learning curve values are unavailable in the saved model log.")
}
```

Lower validation log loss indicates better probability performance under
this metric. Lower training loss alone does not establish generalisation.

# Validation Threshold Selection

Thresholds from 0.01 to 0.99 are evaluated on the validation set.
The highest missed-appointment F1-score determines each selected threshold.

```{r thresholds}
knitr::kable(
  report$validation_results,
  digits = 4,
  row.names = FALSE
)
```

Random Forest uses threshold **`r report$rf_threshold`**.
XGBoost uses threshold **`r report$xgb_threshold`**.

These are follow-up thresholds. They can flag an appointment even when
attending remains the more probable outcome.

# Test Evaluation

The models are evaluated on appointments from patients excluded from
model training and validation.

## Evaluation Metrics

- **Accuracy:** proportion of all appointments classified correctly.
- **Precision:** proportion of flagged appointments that were missed.
- **Recall:** proportion of actual missed appointments identified.
- **F1-score:** harmonic mean of precision and recall.
- **Brier score:** mean squared probability error; lower is better.

## Performance Comparison

```{r comparison-table}
knitr::kable(
  report$comparison,
  digits = 4,
  row.names = FALSE
)
```

```{r comparison-chart}
metric_values <- t(as.matrix(
  report$comparison[, c("Accuracy", "Precision", "Recall", "F1")]
))

colnames(metric_values) <- report$comparison$Model

barplot(
  metric_values,
  beside = TRUE,
  col = c("steelblue", "darkorange", "seagreen", "mediumpurple"),
  ylim = c(0, 1.15),
  main = "Test Performance Comparison",
  ylab = "Score",
  legend.text = rownames(metric_values),
  args.legend = list(
    x = "topright",
    bty = "n",
    cex = 0.8
  )
)
```

## Confusion Matrices

Rows represent actual outcomes. Columns represent threshold-based
predictions.

```{r confusion-matrices}
rf_cm <- table(
  Actual = report$actual,
  Predicted = report$rf_prediction
)

xgb_cm <- table(
  Actual = report$actual,
  Predicted = report$xgb_prediction
)

knitr::kable(
  as.matrix(rf_cm),
  caption = "Random Forest Confusion Matrix"
)

knitr::kable(
  as.matrix(xgb_cm),
  caption = "XGBoost Confusion Matrix"
)
```

## Baseline Performance

The always-attend baseline predicts attendance for every appointment.
The probability baseline uses the model-training missed proportion
for every test appointment.

```{r baseline}
actual_binary <- as.integer(report$actual == "Missed")

knitr::kable(data.frame(
  Metric = c(
    "Always-attend test accuracy",
    "Constant-probability test Brier score"
  ),
  Value = c(
    mean(actual_binary == 0),
    mean((report$baseline_probability - actual_binary)^2)
  )
), digits = 4)
```

Lower thresholds can improve recall while reducing accuracy.
False positive counts indicate the number of unnecessary follow-up flags.

## Results Interpretation

```{r findings, results='asis'}
for (i in seq_len(nrow(report$comparison))) {
  result <- report$comparison[i, ]

  cat(sprintf(
    "- **%s:** precision %.2f%%, recall %.2f%%, F1 %.4f, Brier score %.4f.\n\n",
    result$Model,
    100 * result$Precision,
    100 * result$Recall,
    result$F1,
    result$BrierScore
  ))
}
```

Small differences between model scores do not establish a statistically
reliable winner. Model selection should consider follow-up capacity,
missed cases, and probability quality.

# Feature Importance

## Random Forest

```{r rf-importance}
importance <- sort(report$rf_importance)

old_par <- par(mar = c(5, 10, 4, 2))

barplot(
  importance,
  horiz = TRUE,
  las = 1,
  col = "steelblue",
  main = "Random Forest Feature Importance",
  xlab = "Impurity Importance"
)

par(old_par)
```

## XGBoost

```{r xgb-importance}
knitr::kable(report$xgb_importance, digits = 4)

old_par <- par(mar = c(5, 10, 4, 2))

barplot(
  rev(report$xgb_importance$Gain),
  names.arg = rev(report$xgb_importance$Feature),
  horiz = TRUE,
  las = 1,
  col = "seagreen",
  main = "XGBoost Feature Importance",
  xlab = "Gain"
)

par(old_par)
```

Importance describes model usage, not causation. The two importance
methods use different scales.

# New Patient Prediction

The following patient details are fictional. The actual attendance
outcome is unknown.

```{r new-patient}
patient <- report$new_patient
patient$Gender <- as.character(patient$Gender)

knitr::kable(patient)
```

## Attending and Missing Probabilities

```{r prediction-table}
knitr::kable(
  report$new_patient_results,
  row.names = FALSE
)
```

```{r prediction-chart}
prediction <- report$new_patient_results

barplot(
  rbind(
    Attending = prediction$Attending_Percent,
    Missing = prediction$Missing_Percent
  ),
  names.arg = prediction$Model,
  col = c("seagreen", "tomato"),
  ylim = c(0, 115),
  main = "New Patient Attendance Probabilities",
  ylab = "Estimated Probability (%)",
  legend.text = c("Attending", "Missing"),
  args.legend = list(
    x = "topright",
    bty = "n",
    cex = 0.8
  )
)
```

Attending and missing probabilities total approximately 100%, subject
to rounding. The more probable outcome uses a 50% threshold.
The follow-up flag uses the validation-selected threshold.

# Hospital Management Use

The predictions could help staff prioritise confirmation calls and
reminders. Attendance summaries can support planning discussions.

The experiment does not establish that these actions reduce missed
appointments. A follow-up flag should not automatically cancel an
appointment or restrict patient access.

# Limitations

- The working CSV may contain only part of the original dataset.
- Historical patterns may differ across hospitals and time periods.
- Transport difficulties, urgency, and other relevant factors are absent.
- SMS status does not establish that a message was read.
- SMS status must be available when predictions are made.
- F1-selected thresholds do not reflect actual reminder costs or capacity.
- Probability calibration requires further assessment.
- The test set was inspected during development; a fresh external
  evaluation would provide stronger evidence.
- A fictional patient example demonstrates usage, not model validity.

# Conclusion

The project implements data preparation, attendance analysis, and
probability prediction using Random Forest and XGBoost in R.
Patient-separated partitions support evaluation on unseen patients.
Validation-selected thresholds define follow-up flags, while both
attending and missing probabilities remain visible.

The implementation is a research prototype. Practical use requires
stronger validation and assessment of reminder workload and benefits.

# Future Work

- Verify and document source dataset completeness.
- Evaluate on later appointments and external hospital records.
- Assess calibration and performance across patient groups.
- Choose thresholds using hospital follow-up capacity.
- Add an interface with patient input validation.
- Evaluate whether targeted reminders improve actual attendance.

# References

1. Medical Appointment No Shows dataset, supplied working file
   `KaggleV2-May-2016.csv`.
2. `ranger` R package documentation.
3. `xgboost` R package documentation.
4. `rmarkdown` R package documentation.
5. `knitr` R package documentation.

# Software Environment

```{r software-environment}
cat(report$software_environment, sep = "\n")
```
)---"

writeLines(
  report_text,
  file.path(report_dir, "Hospital_Attendance_Report.Rmd")
)

cat("Complete report source created successfully.\n")
````
